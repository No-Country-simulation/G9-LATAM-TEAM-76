# Módulo 06 — Persistencia y configuración

## Objetivo

Este módulo consolida la configuración y los artefactos que permiten recuperar el pipeline sin repetir cálculos innecesarios. Centralizar estos valores evita que notebooks diferentes utilicen modelos, métricas o dimensiones incompatibles.

## Artefactos administrados

El directorio `models/` contiene `dataset_procesado.json`, `documents.json`, `embeddings.npy`, `semantic_index.joblib` y `model_config.json`. Los primeros módulos producen los datos, embeddings e índice; este módulo registra cómo fueron producidos.

## Configuración registrada

`model_config.json` incluye el nombre del modelo preentrenado, `normalize_embeddings`, la métrica coseno, el número de documentos y la dimensión de los embeddings. En el prototipo se esperan 900 documentos y dimensión 384.

## Relación con una futura API

Una API futura podrá cargar estos artefactos al iniciar y llamar a la función de búsqueda sin descargar nuevamente el dataset ni reconstruir el índice. Esta separación mantiene el pipeline reproducible y facilita validar que una consulta utiliza exactamente la misma configuración que el corpus.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd() / 'src'))
from techmind.config import get_paths, ensure_directories
from techmind.persistence.artifacts import load_documents, load_embeddings, save_model_config
paths = get_paths(); ensure_directories(paths)
documents = load_documents(paths.documents); embeddings = load_embeddings(paths.embeddings)
save_model_config(paths.model_config, len(documents), embeddings.shape[1])
print(f'[TechMind] Configuration persisted at {paths.model_config}')
```
