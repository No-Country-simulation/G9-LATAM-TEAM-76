# Módulo 05 — Evaluación semántica multilingüe

## Objetivo

Este módulo evalúa si los embeddings representan de forma parecida documentos que realmente están relacionados entre inglés, español y portugués. No crea etiquetas artificiales: utiliza `original_url` como relación observada en los datos.

## ¿Qué es un documento paralelo?

Un grupo paralelo contiene versiones lingüísticas de un mismo contenido. Para cada grupo completo se comparan los pares EN-ES, EN-PT y ES-PT mediante `cosine_similarity`.

## Entrada

Se utilizan `models/dataset_procesado.json` y `models/embeddings.npy`. Es indispensable que el orden de las filas sea idéntico, porque cada embedding debe corresponder a su documento.

## Salida

Se produce una tabla con `original_url`, `id_a`, `idioma_a`, `id_b`, `idioma_b` y `similitud`, además de estadísticas `count`, `mean`, `std`, `min`, percentiles y `max` separadas por par lingüístico.

## Interpretación

Una similitud alta sugiere que dos documentos ocupan posiciones cercanas en el espacio semántico; una similitud baja puede reflejar diferencias de contenido, ruido, traducciones incompletas o limitaciones del modelo. La similitud no demuestra equivalencia perfecta.

## Qué no hace este módulo

No utiliza accuracy, F1, train/test split de clasificación, categorías ni modelos supervisados. Es una evaluación basada en relaciones explícitas del dataset y debe interpretarse junto con una inspección cualitativa.

```python
import sys
from pathlib import Path
import numpy as np
sys.path.insert(0, str(Path.cwd() / 'src'))
from techmind.config import get_paths
from techmind.persistence.artifacts import load_documents, load_embeddings
from techmind.evaluation.multilingual import evaluate_parallel_documents, descriptive_statistics
paths = get_paths(); documents = load_documents(paths.dataset); embeddings = load_embeddings(paths.embeddings)
scores = evaluate_parallel_documents(documents, embeddings)
stats = descriptive_statistics(scores)
scores.to_csv(paths.reports / 'evaluation' / 'multilingual_scores.csv', index=False)
stats.to_csv(paths.reports / 'evaluation' / 'multilingual_statistics.csv')
display(stats)
```
