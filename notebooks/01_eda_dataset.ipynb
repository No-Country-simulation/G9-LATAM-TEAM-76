# Módulo 01 — EDA y preparación del dataset

## Objetivo

Este módulo carga exclusivamente `AI-Culture-Commons/ai-culture-multilingual-json-dolma`, inspecciona su estructura y prepara un subconjunto reproducible para el prototipo semántico de TechMind. El análisis incluye valores nulos, duplicados, distribución por idioma y longitudes de títulos y textos.

## Problema que resuelve

Los datos originales pueden contener idiomas, formatos Unicode, espacios, campos vacíos o registros duplicados que dificulten las etapas posteriores. Aquí se aplica una limpieza conservadora: Unicode NFC, eliminación de caracteres de control, normalización de espacios y conservación de acentos. No se convierte el texto a ASCII.

## Entrada

Se utiliza la configuración `json` y el split `train`. Se conservan exclusivamente los idiomas `en`, `es` y `pt`. Los campos `title`, `content` y `language` se renombran como `titulo`, `texto` e `idioma`.

## Selección paralela

`original_url` funciona como vínculo explícito entre versiones relacionadas. Se seleccionan 300 URLs que contienen EN, ES y PT, con una fila por idioma, para producir 900 documentos reproducibles usando `random_state=42`.

## Salidas

Se generan `models/dataset_procesado.json` y `models/documents.json`, ambos con `id`, `titulo`, `texto`, `idioma`, `url` y `original_url`. Estos archivos son la entrada del Módulo 02.

## Decisiones y limitaciones

La limpieza conserva información lingüística y no inventa etiquetas ni categorías. El subconjunto de 900 documentos es adecuado para demostración, pero no representa necesariamente todo el dataset original.

```python
import sys
from pathlib import Path
# import matplotlib.pyplot as plt
sys.path.insert(0, str(Path.cwd() / 'src'))
from techmind.config import ensure_directories, get_paths, PROTOTYPE_PER_LANGUAGE
from techmind.data.preprocessing import load_source_dataset, prepare_documents, select_parallel_prototype
from techmind.persistence.artifacts import save_documents
paths = get_paths(); ensure_directories(paths)
raw = load_source_dataset()
documents = prepare_documents(raw)
prototype = select_parallel_prototype(documents, PROTOTYPE_PER_LANGUAGE)
save_documents(prototype, paths.dataset)
save_documents(prototype, paths.documents)
display(prototype.groupby('idioma').size())
display(prototype.assign(words=prototype.texto.str.split().str.len()).describe(include='all'))
print("Completado")
```
